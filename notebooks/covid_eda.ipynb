### Case Analysis

```python
total_cases_max = df.groupby("country")["total_cases"].max()

total_cases_max.sort_values(ascending=False).head(15)
# total_cases_max["India"]
Cases_per_country=(df.groupby("country").agg(
    Total_Cases = ("total_cases","max"),
    Max_New_Cases= ("new_cases","max")
).reset_index())
peak_dates=df.loc[
    df.groupby("country")["new_cases"].idxmax().dropna(), 
    ["country","date"]
]
peak_dates=peak_dates.rename( columns={"date":"max_newcase_date"})
# used dropna bcz peak_dates was getting some NAN values
Cases_per_country=Cases_per_country.merge(peak_dates,on="country")
# Cases_per_country[Cases_per_country["country"]=="India"]
Cases_per_country.sort_values(by="Max_New_Cases",ascending=False).head(10)
```

```python
country_cases=Cases_per_country[~Cases_per_country["country"].str.contains("World|excl.|Europe|Asia|North|South|income|Upper|European")]
country_cases.sort_values(by="Total_Cases",ascending=False)
```

### Death Analysis

```python
df_deaths=df.groupby("country").agg(
    total_deaths=("total_deaths","max"),
    new_deaths_max=("new_deaths","max")
).reset_index().sort_values(by="new_deaths_max",ascending=False)

peak_death_date=df.loc[df.groupby("country")["new_deaths"].idxmax().dropna(),("country","date")]
df_deaths=df_deaths.merge(peak_death_date,on="country")
df_deaths

df_deaths=df_deaths[~df_deaths["country"].str.contains("World|excl.|Europe|Asia|North|South|income|Upper|European")]

df_deaths
# we will merge this table with last country_cases table to get complete analysis of cases and death all together and extract important information out
df_final_case_death=pd.merge(country_cases,df_deaths,on="country",how="left")
df_final_case_death["case_fatality_rate"]=np.round((df_final_case_death["total_deaths"]*100/df_final_case_death["Total_Cases"]),4)
df_final_case_death.sort_values(by="Total_Cases",ascending=False)
df_final_case_death.sort_values(by="new_deaths_max",ascending=False).head(10)
```

```python
df_final_case_death.sort_values(by="case_fatality_rate",ascending=False).head(10)
```

- **Observations**

    * Countries with a larger number of cumulative COVID-19 cases did not always experience the highest daily death counts. This suggests that total infections alone cannot fully explain the severity of mortality across countries.
    
    * In most countries, the peak in daily deaths occurred after the peak in daily cases. This time gap indicates that the impact of a surge in infections is often reflected in mortality statistics after a delay.
    
    * China recorded the highest single-day death count in the dataset, reporting **44,047 deaths on 2023-02-01**. Other countries with notably high peak daily deaths include **Chile, Japan, Ecuador, Germany, and India**.
    
    * India reported its highest daily death count of **6,148 deaths on 2021-06-10**, which coincides with the period following the severe second wave of COVID-19 in the country.
    
    * The dates of maximum daily cases and maximum daily deaths are generally different, highlighting that infection waves and mortality waves do not peak simultaneously.
    
    * Peak daily death counts provide useful insight into the intensity and impact of COVID-19 waves experienced by different countries during the pandemic.
    
    * Overall, the analysis shows that while case counts are an important indicator of disease spread, mortality trends are influenced by multiple factors and may follow a different pattern over time.
 
    * **Yemen** recorded the highest case fatality rate in the dataset (18.07%), making it a clear outlier compared to other countries. This may reflect factors such as limited testing capacity, under-reporting of infections, healthcare challenges, or differences in reporting practices.

Countries that conducted widespread testing generally reported lower positivity rates, while countries with limited testing coverage often experienced higher positivity rates.

This suggests that while increased testing may contribute to lower positivity rates, other factors such as outbreak severity, population characteristics, government policies, and reporting practices also played an important role.
    
    * In India, the highest positivity rate (22.3%) was recorded on 11 May 2021, only a few days after the peak in daily new cases on 7 May 2021. This coincides with the country's severe second wave and reflects the period of intense community transmission.

- *Observation*

    - China recorded the highest number of total vaccination doses administered globally, followed by India, Africa, the United States, and Brazil. This reflects the large-scale vaccination campaigns carried out in highly populated regions.
    
    - China also achieved the highest peak in daily vaccinations, administering more than **22 million doses (smoothed value)** on **28 June 2021**. India recorded the second-highest peak, reaching over **10 million doses** on **24 September 2021**.
    
    - Population-adjusted metrics tell a different story. Countries such as **Gibraltar, Cook Islands, Qatar, Brunei, and Nauru** recorded the highest percentages of fully vaccinated individuals, demonstrating that smaller countries were able to achieve broader vaccination coverage relative to their population size.
 
    -  A few countries reported `people_vaccinated_per_hundred` values above 100. While this is theoretically impossible if only residents are considered, it can occur due to differences in population estimates, vaccination of non-residents, temporary workers, cross-border populations, or reporting methodologies. Such values are more common in countries with relatively small populations.
    
    - The timing of vaccination rollouts varied significantly across countries. China reached its peak vaccination activity earlier than India, indicating a faster acceleration of its vaccination campaign during the initial rollout phase.
    
    - The earliest vaccination records in the dataset appear in **December 2020**, marking the beginning of mass vaccination efforts shortly after COVID-19 vaccines became available.
    
    - India's first recorded vaccination entry occurred on **16 January 2021**, with approximately **191 thousand doses** administered on that day. This marked the launch of one of the largest vaccination campaigns in the world.
    
    - Booster dose administration was concentrated in a few countries. China, India, Japan, the United States, and Brazil reported the highest total number of booster doses, reflecting continued vaccination efforts beyond the primary vaccination schedule.

    - A substantial proportion of vaccination-related variables contain missing values, with several columns having more than **80% missing data**. Therefore, comparisons across countries should be interpreted carefully, as reporting practices and data availability differ considerably.
 
    - Despite extensive analysis, no strong negative relationship was observed between vaccination coverage and COVID-19 death rates across countries. The scatter plots and correlation analysis suggest that vaccination alone is not sufficient to explain variations in mortality, indicating that other factors such as healthcare infrastructure, demographics, government policies, virus variants, and reporting practices also played significant roles in influencing death rates during the pandemic.
