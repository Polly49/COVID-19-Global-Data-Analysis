### Case Analysis

```python
total_cases_max = df.groupby("country")["total_cases"].max()

total_cases_max.sort_values(ascending=False).head(15)
# total_cases_max["India"]
Cases_per_country=(df.groupby("country").agg(
    Total_Cases = ("total_cases","max"),
    Max_New_Cases= ("new_cases","max")
).reset_index())
peak_dates=df.loc[
    df.groupby("country")["new_cases"].idxmax().dropna(), 
    ["country","date"]
]
# used dropna bcz peak_dates was getting some NAN values
Cases_per_country=Cases_per_country.merge(peak_dates,on="country")
# Cases_per_country[Cases_per_country["country"]=="India"]
Cases_per_country.sort_values(by="Max_New_Cases",ascending=False).head(10)
```

```python
Cases_per_country[Cases_per_country["country"].isin(["China","India"])]
```

```python
country_cases=Cases_per_country[~Cases_per_country["country"].str.contains("World|excl.|Europe|Asia|North|South|income|Upper|European")]
country_cases.sort_values(by="Total_Cases",ascending=False)
```

```python
print(df["new_cases_smoothed"].max())
print(df["new_cases"].max())
df[["date","new_cases","new_cases_smoothed"]].sort_values(by="new_cases",ascending=False).head(8)
```

- *Obsevation*
    - Highest case in a day reported by any Individual country is China and which is On 2022-12-2023
    - Total cases in India was 45056221 and maximum new_cases found in a day in India was 414188 On 2021-05-07
    - In dataset(df) there is a data of different continents which will always be more then individual countries so for different country comparison we need to filter out countries.
    - United States had highest number of cases followed by China and India.(It shows that USA,China and India were most affected)
    - Smoothed features shows the seven days cummulative avg of all the new cases such that it become smoothed new cases.Because Covid do not exploded suddenly so we can get better trend analysis with this smoothed data
